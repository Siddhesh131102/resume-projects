```python
# Walmart Sales Data Analysis using Python (Pandas)
```

```python
# Importing Pandas

import pandas as pd
```

```python
# Loading dataset

df=pd.read_csv('Walmart_Sales.csv')
```

```python
pd.options.display.float_format = '{:,.2f}'.format
```

```python
# Preview data

df.head()
```

```python
# Checking dataset structure

df.info()
```

```python
# Statistical summary

df.describe()
```

```python
# Converting date column to datetime

df['Date']=pd.to_datetime(df['Date'],dayfirst=True)
```

```python
df.info()
```

```python
# Filling missing temperature

df['Temperature']=df['Temperature'].fillna(df['Temperature'].median())
```

```python
# Filling missing fuel price

df['Fuel_Price']=df['Fuel_Price'].fillna(df['Fuel_Price'].median())
```

```python
# Extracting month from date

df['Month']=df['Date'].dt.month
```

```python
# # Extracting year from date

df['Year']=df['Date'].dt.year
```

```python
df
```

```python
# Total sales by store

store_sales_2012=df[df['Year']==2012].groupby('Store')['Weekly_Sales'].sum().sort_values(ascending=False)
```

```python
store_sales_2012
```

```python
df
```

```python
# Filtering data for Store 33

df[df['Store']==33]
```

```python
# Monthly sales trend

Monthly_sales=df.groupby(['Year','Month'])['Weekly_Sales'].sum()
```

```python
Monthly_sales
```

```python
# Yearly sales

yearly_sales=df.groupby('Year')['Weekly_Sales'].sum()
```

```python
yearly_sales
```

```python
# Holiday vs Non-Holiday sales

holiday_sales=df.groupby('Holiday_Flag')['Weekly_Sales'].mean()


# Average weekly sales during holidays are higher than non-holiday periods (1.04M), indicating that sales tend to increase during holiday periods.
```

```python
holiday_sales
```

```python
# Pivot table

monthly_sales_table=df.pivot_table(values='Weekly_Sales',index='Month',columns='Year',aggfunc='sum')
```

```python
monthly_sales_table
```

```python
# Calculating sales difference between 2012 and 2011

monthly_sales_table['2012_vs_2011']=monthly_sales_table[2012]-monthly_sales_table[2011]
```

```python
monthly_sales_table
```

```python
# Calculating year-over-year percentage change (2012 vs 2011)


monthly_sales_table['YoY_percent']=(((monthly_sales_table[2012]-monthly_sales_table[2011])/monthly_sales_table[2011])*100)


# Year-over-year analysis shows that March, June, and August experienced the highest growth, indicating these months contributed significantly to overall sales growth.
```

```python
monthly_sales_table.sort_values(by='YoY_percent',ascending=False)
```

```python
# Sorting months by sales difference between 2012 and 2011


monthly_sales_table.sort_values(by='2012_vs_2011',ascending=False)


# Month-wise comparison between 2012 and 2011 shows that March, June, and August recorded the highest growth, with sales increasing by over 48M, making them the top contributors to year-over-year sales improvement.
```

```python
df
```

```python
# Sorting monthly sales by 2012 in descending order


monthly_sales_table.sort_values(by=2012,ascending=False)


# Monthly sales analysis shows that June, August, and March 2012 recorded the highest sales, each exceeding 230M, with strong year-over-year growth of 25–29%.
```

```python
# Correlation between CPI and Sales


df[['CPI','Weekly_Sales']].corr()


# As CPI increases, sales slightly decrease, but the effect is almost negligible.
```

```python
# Correation between unemployment and sales


df[['Unemployment','Weekly_Sales']].corr()


# There is a weak negative correlation (-0.11) between unemployment rate and weekly sales, suggesting that higher unemployment may slightly reduce sales, but the relationship is not strong enough to be a key driving factor.
```

```python
# Correlation between fuel price and sales


df[['Fuel_Price','Weekly_Sales']].corr()


# There is almost no correlation between fuel prices and weekly sales,indicating fuel price fluctuations do no significantly influence sales trends.
```
